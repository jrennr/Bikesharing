```python
from pathlib import Path
import pandas as pd

file_path = Path('201908-citibike-tripdata.csv')
citibike_data = pd.read_csv(file_path)
citibike_data.head()
```

```python
citibike_data.dtypes
```

```python
citibike_data['tripduration'] = pd.to_datetime(pd.to_numeric(citibike_data['tripduration'], errors='raise'), unit='s')
citibike_data.dtypes
```

```python
output_path = file_path.with_name(file_path.stem + '_converted.csv')
citibike_data.to_csv(output_path, index=False)
print(output_path)
```
